```python
def set_global_seed(seed: int) -> None:
    """Set global seed for reproducibility.
    :param int seed: Seed to be set
    """
    torch.backends.cudnn.deterministic = True
    torch.manual_seed(seed)
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    np.random.seed(seed)
    random.seed(seed)

    # также можно зафиксировать seed для Dataloader
    g = torch.Generator()
    g.manual_seed(seed)
    return g

# Сид для каждого worker в Dataloader
def seed_worker(worker_id):
    worker_seed = torch.initial_seed() % 2**32
    np.random.seed(worker_seed)
    random.seed(worker_seed)
SEED = 52
set_global_seed(SEED)
seed_worker(SEED)
```

```python
# your code here (ง •̀_•́)ง
class LinRegModel(nn.Module):
    def __init__(self, num_params):
        super().__init__()
        self.lin_sloy = nn.Linear(in_features=num_params, out_features=1)

    def forward(self, x):
        return self.lin_sloy(x)
```

```python
X_test_tensor = torch.tensor(X_test, dtype=torch.float32, device=device)
y_test_tensor = torch.tensor(y_test, dtype=torch.float32, device=device)
X_train_tensor = torch.tensor(X_train, dtype=torch.float32, device=device)
y_train_tensor = torch.tensor(y_train, dtype=torch.float32, device=device)
X_val_tensor = torch.tensor(X_val, dtype=torch.float32, device=device)
y_val_tensor = torch.tensor(y_val, dtype=torch.float32, device=device)
```

```python
def test_model(model, X_test, y_test, batch_size=300):
    model.eval()
    y_test = y_test.reshape(-1, 1)
    tot_error = 0.0
    n = X_test.shape[0]
    with torch.no_grad():
        for start in range(0, n, batch_size):
            X = X_test[start: start + batch_size].to(device=device, dtype=torch.float32)
            y = y_test[start: start + batch_size].to(device=device, dtype=torch.float32)

            pred = model(X)

            tot_error += ((pred - y) ** 2).sum().item()
    return tot_error / n
```

```python

def train_model(model, X_train, y_train, X_val, y_val, batch_size, learning_rate=0.01, epochs=10):
    y_train = y_train.reshape(-1, 1)
    n = X_train.shape[0]
    model.train()
    val_errors = []
    train_errors = []
    for epoch in range(epochs):
        ind = torch.randperm(n, device=X_train.device)

        for start in range(0, n, batch_size):
            ind_batch = ind[start: start + batch_size]
            X = X_train[ind_batch]
            y = y_train[ind_batch]

            model.zero_grad()

            pred = model(X)
            loss = (((pred - y) ** 2).mean())

            loss.backward()

            with torch.no_grad():
                for param in model.parameters():
                    param -= learning_rate * param.grad
        val_err = test_model(model, X_val, y_val) ** 0.5
        train_err = test_model(model, X_train, y_train) ** 0.5

        val_errors.append(val_err)
        train_errors.append(train_err)

        clear_output(wait=True)

        num_epoch = range(1, epoch + 2)

        plt.figure(figsize=(10, 6))
        plt.plot(num_epoch, val_errors, label='Val')
        plt.plot(num_epoch, train_errors, label='Train')

        plt.xlabel("Эпоха")
        plt.ylabel('RMSE')
        plt.legend()
        plt.grid()
        plt.show()
        plt.close()



```

## Часть II. Заводим нейронную сеть (максимум 10 баллов)

Ниже нам предстоит реализовать довольно много различных нейросетей и поставить целую серию экспериментов. Чтобы это всё происходило без боли и страданий, нам нужно держать код в удобном виде.

При решении заданий вы можете придерживаться любой адекватной струкуры кода, но мы советуем воспользоваться сигнатурами функций, которые приведены ниже. При необходимости вы можете добавить в них любые нужные вам аргументы и любой нужный функционал. Более того, хорошей практикой является не делать эти функции слишком громоздкими и выносить разные хитрые штуки в отдельные функции.

```python
def prepare_batch(batch, task):
    batch_x = batch['sample'].to(device=device, dtype=torch.float32)
    batch_y = batch['target'].to(device=device)

    if task == 'class':
        batch_y = batch_y.long().reshape(-1)
    else:
        batch_y = batch_y.float().reshape(-1, 1)
    return batch_x, batch_y
```

```python
def training_epoch(model, optimizer, criterion, train_loader, num_epoch, task):
    """Одна эпоха обучения
    params:
        model - torch.nn.Module to be fitted
        optimizer - model optimizer
        criterion - loss function from torch.nn
        train_loader - torch.utils.data.Dataloader with train set
    """

    # your code here  ♪┏(・o･)┛♪
    model.train()
    train_loss = 0.0
    num_sample = 0
    for batch in tqdm(train_loader, desc=f"Обучение: эпоха №{num_epoch}"):
        batch_x, batch_y = prepare_batch(batch, task)


        optimizer.zero_grad()

        pred = model(batch_x)
        loss = criterion(pred, batch_y)

        loss.backward()
        optimizer.step()

        size_batch = batch_x.shape[0]
        train_loss += loss.item() * size_batch
        num_sample += size_batch

    return train_loss / num_sample




@torch.no_grad()
def validation_epoch(model, criterion, val_loader, task):
    """Одна эпоха валидации модели
    params:
        model - torch.nn.Module to be fitted
        criterion - loss function from torch.nn
        val_loader - torch.utils.data.Dataloader with test set
                      (if you wish to validate during training)
    """

    # your code here   ฅ^•ﻌ•^ฅ
    model.eval()
    val_loss = 0.0
    num_sample = 0
    for batch in tqdm(val_loader, desc=f"Валидация"):
        batch_x, batch_y = prepare_batch(batch, task)

        pred = model(batch_x)
        loss = criterion(pred, batch_y)

        size_batch = batch_x.shape[0]
        val_loss += loss.item() * size_batch
        num_sample += size_batch

    return val_loss / num_sample






@torch.no_grad()
def predict(model, data_loader, task):
    """ Предсказания модели
    params:
        model - torch.nn.Module to be evaluated on test set
        criterion - loss function from torch.nn
        data_loader - torch.utils.data.Dataloader with test set
    ----------
    returns:
        predicts - torch.tensor with shape (len(test_loader.dataset), ),
                   which contains predictions for test objects
    """

    # your code here  =^･ω･^=
    model.eval()
    predicts = []
    for i, batch in enumerate(data_loader):
        batch_x = batch['sample'].to(device, dtype=torch.float32)
        pred = model(batch_x)
        if task == 'class':
            pred = pred.argmax(dim=1)
        predicts.append(pred)


    return torch.cat(predicts, dim=0)


def train(model, optimizer, criterion, train_loader, val_loader, epochs, task):
    """ Обучение модели
    params:
        model - torch.nn.Module to be fitted
        optimizer - model optimizer
        criterion - loss function from torch.nn
        train_loader - torch.utils.data.Dataloader with train set
        val_loader - torch.utils.data.Dataloader with test set
                      (if you wish to validate during training)
        epochs - number of training epochs
    """

    # your code here  ¯\_(ツ)_/¯
    val_errors = []
    train_errors = []

    for epoch in range(epochs):
        training_epoch(model, optimizer, criterion, train_loader, epoch+1, task)
        val_err = validation_epoch(model, criterion, val_loader, task)
        train_err = validation_epoch(model, criterion ,train_loader, task)

        val_errors.append(val_err)
        train_errors.append(train_err)

        clear_output(wait=True)

        num_epoch = range(1, epoch + 2)

        plt.figure(figsize=(10, 6))
        plt.plot(num_epoch, val_errors, label='Val')
        plt.plot(num_epoch, train_errors, label='Train')

        plt.xlabel("Эпоха")
        plt.ylabel('Loss')
        plt.legend()
        plt.grid()
        plt.show()
        plt.close()


```

```python
class RandomDataset(torch.utils.data.Dataset):

    def __init__(self, x, y):
        self.x=x
        self.y=y

    def __len__(self):
        return len(self.x)

    def __getitem__(self, idx):
        return {'sample':self.x[idx, :], 'target': self.y[idx]}
```

**Задание 6 (1 балл).** Начнем с того, что попробуем отнормировать целевую переменную. Для этого воспользуемся min-max нормализацией, чтобы целевая переменная принимала значения от 0 до 1. Реализуйте функции `normalize` и `denormalize`, которые, соответственно, нормируют целевую переменную и применяют обратное преобразование. Минимум и максимум оцените по обучающей выборке (то есть эти константы должны быть фиксированными и не зависеть от передаваемой выборки).

```python
y_min = y_train_tensor.min().item()
y_max = y_train_tensor.max().item()
```

```python

def normalize(sample):
    """
    Min-max normalization to convert sample to [0, 1] range
    """
    # your code here ᕦ(ò_óˇ)ᕤ
    return (sample - y_min) / (y_max - y_min)



def denormalize(sample):
    """
    Denormalize sample from [0, 1] to initial range
    """
    # your code here ( ⚆ ω ⚆)
    return sample * (y_max - y_min) + y_min
```
